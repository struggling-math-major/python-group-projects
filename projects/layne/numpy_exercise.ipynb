TL;DR , lists are ordered and changeable.

__Ordered:__ 

we can access specific elements in a list based on their index

````python
# make an empty list
listy = []
# make a list with various values of different types
listy2 = [3 , "apple", 1]
# access the first value in a list
listy2[0]
# access the last value in a list
listy2[-1]
# access a portion of the list from i to j-1
listy2[0:2]
# print your list to check
print(listy2)
````

__Changeable__: 

we can add and remove things from a list-- mostly for our example we are concerned with adding elements

````python
# append an object to the end of a list
listy2.append("a new object")
````
