## our dataset

two fictional numerical methods estimate the same quantity, whose true value is 10.

each row represents one run:
- method: which method was used
- difficulty: easy or hard problem
- estimate: the computed answer
- runtime: time taken, in seconds

run the next cell to create our table.

```python
# fictional results, chosen for our comparison puzzle
results = pd.DataFrame({
    "method": ["A"] * 10 + ["B"] * 10,
    "difficulty": (
        ["easy"] * 2 + ["hard"] * 8
        + ["easy"] * 8 + ["hard"] * 2
    ),
    "estimate": [
        # method A: 2 easy runs, then 8 hard runs
        10.1, 9.9,
        11.0, 9.0, 11.2, 8.8, 10.8, 9.2, 11.0, 9.0,
        # method B: 8 easy runs, then 2 hard runs
        10.2, 9.8, 10.3, 9.7, 10.1, 9.9, 10.2, 9.8,
        11.5, 8.5,
    ],
    "runtime": [
        0.10, 0.12,
        0.40, 0.42, 0.50, 0.48, 0.35, 0.38, 0.44, 0.43,
        0.08, 0.09, 0.12, 0.11, 0.07, 0.08, 0.10, 0.09,
        0.30, 0.32,
    ],
})

results["true_value"] = 10.0

results
```

## selecting and filtering

a DataFrame is a table with named columns and labeled rows.

- `results["estimate"]` selects one column.
- `results[["method", "estimate"]]` selects multiple columns.
- a condition produces a column of True/False values.
- putting that condition inside `results[...]` keeps matching rows.

run and modify the examples below.

```python
# select two columns
results[["method", "estimate"]]
```

```python
# keep only runs taking less than 0.1 seconds
results[results["runtime"] < 0.1]
```

## grouping and summarizing

`groupby` splits rows into groups. we can then calculate a summary
for each group.

for example:
- `.mean()` calculates the mean.
- `.min()` finds the minimum.
- `.size()` counts the rows in each group.

you can group by one column or a list of columns.

```python
# mean runtime for each difficulty category
results.groupby("difficulty")["runtime"].mean()
```
